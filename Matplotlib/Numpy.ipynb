```python
# Iterating the Array
```

```python
iterate_values = np.array([[[1, 2, 3], [4, 5, 6]],[[12, 13, 14], [15, 16, 17]]])
```

```python
for value in iterate_values:
    print(value)
```

```python
for value in iterate_values:
    for value_1 in value:
        for value_2 in value_1:
            print(value_2)
```

```python
# Iterating array with different data types
```

```python
iterate_values = np.array([[[1, 2, 3], [4, 5, 6]],[[12, 13, 14], [15, 16, 17]]])
```

```python
for value in np.nditer(iterate_values, flags = ['buffered'], op_dtypes = 'S'):
    print(value)
```

```python
# Iterating array with different step size
```

```python
iterate_values = np.array([[[1, 2, 3], [4, 5, 6]],[[12, 13, 14], [15, 16, 17]]])
```

```python
for value in np.nditer(iterate_values[0::, 0]):
    print(value)
```

```python
# enumerate Iteration
```

```python
iterate_values = np.array([[[1, 2, 3], [4, 5, 6]],[[12, 13, 14], [15, 16, 17]]])
```

```python
for id, value in np.ndenumerate(iterate_values):
    print(id,value)
```
