```python
x_axis = np.array((0, 2, 4, 6, 8))
```

```python
y_axis = np.array([1, 3, 5, 6, 9])
```

```python
plt.plot(x_axis, marker = 'o', ls = '-.', c = 'r', linewidth = '0.5')
plt.plot(y_axis, ls = '-', c = 'b', linewidth = '2')

plt.xlabel("time")
plt.ylabel("travelled kms")
plt.title("Distance Travelled")
plt.grid(axis = 'x')
plt.style.available
```

```python
# multiple plots (subplot)
```

```python
plot_1x = np.array([1, 2, 3])
```

```python
plot_1y = np.array([0, 4, 6])
```

```python
plt.subplot(2, 1, 1)
plt.plot(plot_1x, plot_1y)
```

```python
plot_2x = np.array([0, 2, 4, 6, 8])
```

```python
plot_2y = np.array([1, 3, 4, 2, 4])
```

```python
plt.subplot(2, 1, 1)
plt.plot(plot_1x, plot_1y)
plt.title("Absent")

plt.subplot(2, 1, 2)
plt.plot(plot_2x, plot_2y)
plt.title("Present")

plt.suptitle("Attendance")
```

```python
#Scatter
```

```python
plot_1x = np.array([1, 2, 3, 4, 0])
plot_1y = np.array([0, 4, 6, 2, 5])

plot_2x = np.array([0, 2, 4, 6, 8])
plot_2y = np.array([1, 3, 4, 2, 4])

colors = np.array(["red", "blue", "green", "yellow", "orange"])

plt.scatter(plot_1x, plot_1y, c = colors)
plt.colorbar()
plt.scatter(plot_2x, plot_2y,  c = colors)
```
