```python
import pandas as pd
```

```python
Person = { "Id" : [1, 4, 3],
           "Name" : ["Jo", "Bobby", "Amaran"]}
```

```python
first = pd.DataFrame(Person)
```

```python
Person_1 = {"Id" : [1, 4],
            "Name" : ["Thilak", "Bobby"]}
```

```python
second = pd.DataFrame(Person_1)
```

```python
merge_value = pd.merge(first, second, on = "Id", how = "inner")
```

```python
merge_value
```

```python
merge_value = pd.merge(first, second, on = "Id", how = "right")
```

```python
merge_value
```

```python
merge_value = pd.merge(first, second, how = "cross")
```

```python
merge_value
```

```python
merge_value.to_json("mock_data.json")
```

```python
concat_values = pd.concat([first, second], ignore_index = True)
```

```python
concat_values
```

```python
concat_values.to_csv("mock_data.csv")
```
